```python
import pandas as pd
import requests
```

```python
response = requests.get("https://api.artic.edu/api/v1/artworks/search?&&page=1")
```

```python
pd.DataFrame(response.json()['data'])[['_score','id','api_model','title','thumbnail']]
```

```python
df = pd.DataFrame()
    
```

```python
df
```

```python
for i in range(1,101):
    response = requests.get(f'https://api.artic.edu/api/v1/artworks/search?page={i}')
    
    temp_df = pd.DataFrame(response.json()['data'])[
        ['_score','id','api_model','title','thumbnail']
    ]
    
    df = pd.concat([df, temp_df], ignore_index=True)
```

```python
df
```

```python
df.to_csv('artworks.csv')
```
